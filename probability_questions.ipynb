### Basic Probability Questions

```python
#What is the probability of getting a head when a coin is tossed once?
favourable_outcomes = 1
total_outcomes = 2

probability = favourable_outcomes / total_outcomes

print(probability)
```

```python
#What is the probability of getting an even number when a die is rolled once?
total_outcomes = 6
even_outcomes = 0

for i in range(1, 7):
    if i % 2 == 0:
        even_outcomes += 1

probability = even_outcomes / total_outcomes

print(probability)
```

```python
#What is the probability of drawing a red card from a standard deck of 52 cards?
total=52
red=0
for suit in ["Hearts","Diamonds","Clubs","Spades"]:
    if suit=="Hearts" or suit =="Diamonds":
        red+=13
prob=red/total
print(prob)
```

```python
#What is the probability of getting an odd number on a fair die?
total=6
odd=0
for i in range(1,7):
    if i%2!=0:
        odd+=1
prob=odd/total
print(prob)
```

```python
#Two cards are drawn from a deck without replacement. Find the probability 
#that one is red and one is black.
cards = ["Red"] * 26 + ["Black"] * 26

total = 0
count = 0

for i in range(52):
    for j in range(52):
        if i != j:
            total += 1

            if cards[i] != cards[j]:
                count += 1

prob = count / total
print(prob)
```

```python
#A coin is tossed three times. Given at least one head, what is the probability 
#of getting exactly two heads?
coin = ["H", "T"]

total = 0
count = 0

for i in coin:
    for j in coin:
        for k in coin:

            if i == "H" or j == "H" or k == "H":
                total += 1

                if (i == "H") + (j == "H") + (k == "H") == 2:
                    count += 1

prob = count / total
print(prob)
```

```python
#A fair coin is tossed three times. Find the probability distribution of 
#the number of heads.
coin=["H","T"]
total=0

c0=0
c1=0
c2=0
c3=0

for i in coin:
    for j in coin:
        for k in coin:
            total+=1

            heads=0
            if i=="H":
                heads+=1
            if j=="H":
                heads+=1
            if k=="H":
                heads+=1

            if heads==0:
                c0+=1
            elif heads==1:
                c1+=1
            elif heads==2:
                c2+=1
            elif heads==3:
                c3+=1
print("P(X=0):", c0 / total)
print("P(X=1):", c1 / total)
print("P(X=2):", c2 / total)
print("P(X=3):", c3 / total)
```

```python
#A die is rolled. Find the expected value of the outcome.
die = [1, 2, 3, 4, 5, 6]

total = 0

for i in die:
    total += i

expected_value = total / len(die)

print(expected_value)
```

```python
#A fair coin is tossed twice. Find the expected number of heads.
coin=["H","T"]
total=0
sum_heads=0

for i in coin:
    for j in coin:
        total+=1

        heads=0

        if i=="H":
            heads+=1
        if j=="H":
            heads+=1
        sum_heads+=heads
expected=sum_heads/total
print(expected)
```

```python
#A coin is tossed 4 times. Find the probability of getting exactly 2 heads.
coin = ["H", "T"]
total = 0
count = 0

for i in coin:
    for j in coin:
        for k in coin:
            for l in coin:
                total += 1

                heads = 0

                if i == "H":
                    heads += 1
                if j == "H":
                    heads += 1
                if k == "H":
                    heads += 1
                if l == "H":
                    heads += 1

                if heads == 2:
                    count += 1

prob = count / total

print(prob)
```

```python
#A coin is tossed 5 times. Find the probability of getting at least 3 heads.
coin = ["H", "T"]
total = 0
count = 0

for i in coin:
    for j in coin:
        for k in coin:
            for l in coin:
                for m in coin:
                    total += 1

                    heads = 0

                    if i == "H":
                        heads += 1
                    if j == "H":
                        heads += 1
                    if k == "H":
                        heads += 1
                    if l == "H":
                        heads += 1
                    if m=="H":
                        heads+=1

                    if heads >= 3:
                        count += 1

prob = count / total

print(prob)
```

```python
#The probability of winning a game is 0.3. Find the probability of winning 
#exactly twice in 5 games.
n = 5
r = 2
p = 0.3
q = 0.7

combination = 10

prob = combination * (p ** r) * (q ** (n-r))

print(prob)
```

```python
#A coin is biased so that the probability of heads is 0.6. Find the 
#probability of getting 3 heads in 5 tosses.
n = 5
r = 3
p = 0.6
q = 1 - p

ans = 10 * (p ** r) * (q ** (n-r))

print(ans)
```

```python
#The probability of success in a single trial is 0.4. Find the probability 
#of exactly 2 successes in 3 trials.
n = 3
r = 2
p = 0.4
q = 1 - p

ans = 3 * (p ** r) * (q ** (n-r))

print(ans)
```

```python
#A letter is known to have come either from city A or city B. Probability 
#it’s from A is 0.6. If a particular mark appears on 3% of letters from A and 
#2% from B, what is the probability the letter is from A given the mark 
#is found?
a = 0.6
b = 1 - a

mark_a = 0.03
mark_b = 0.02

total = (mark_a * a) + (mark_b * b)

ans = (mark_a * a) / total

print(ans)
```
